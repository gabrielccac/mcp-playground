## 1. Basic search

```python
with PNCPClient() as client:
    page = client.search(q="toner", tipos_documento="edital", tam_pagina=5)

print(f"Total results: {page.total_registros}")
print(f"Pages: {page.total_paginas}")
for r in page.items:
    print(f"  - {r.title[:80]} | {r.orgao_nome} | {r.uf}")
```

## 2. Search with status filter

```python
with PNCPClient() as client:
    page = client.search(
        q="toner",
        tipos_documento="edital",
        status="recebendo_proposta",
        tam_pagina=5,
    )

print(f"Open tenders: {page.total_registros}")
for r in page.items:
    print(f"  - {r.title[:80]} | status: {r.situacao_nome}")
```

## 3. Search with modalidade + UF

```python
with PNCPClient() as client:
    page = client.search(
        q="toner hp",
        tipos_documento="edital",
        modalidades=6,   # Pregão Eletrônico
        ufs="SP",
        tam_pagina=5,
    )

print(f"Pregão Eletrônico / SP: {page.total_registros}")
for r in page.items:
    print(f"  - {r.title[:80]} | {r.municipio_nome}")
```

## 4. Raw response inspection (to check field names)

```python
import httpx, json

resp = httpx.get(
    "https://pncp.gov.br/api/search/",
    params={"q": "toner", "tipos_documento": "edital", "pagina": 1, "tam_pagina": 1},
    headers={"User-Agent": "Mozilla/5.0"},
)
data = resp.json()
print("Top-level keys:", list(data.keys()))
if data.get("items"):
    print("\nFirst item keys:", list(data["items"][0].keys()))
    print("\nFirst item:")
    print(json.dumps(data["items"][0], indent=2, ensure_ascii=False))
```
