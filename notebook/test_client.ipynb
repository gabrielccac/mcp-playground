```python
import sys, json
sys.path.insert(0, '..')

from api import search
```

## 1. Basic search

```python
page = search(q="toner", tam_pagina=5)

print(f"Total results : {page['total_registros']}")
print(f"Total pages   : {page['total_paginas']}")
for r in page['items']:
    print(f"  [{r.get('uf')}] {r.get('title','')[:80]} | {r.get('orgao_nome')}")
```

## 2. Status filter — open tenders only

```python
page = search(q="toner", status="recebendo_proposta", tam_pagina=5)

print(f"Open tenders: {page['total_registros']}")
for r in page['items']:
    print(f"  {r.get('title','')[:80]} | {r.get('situacao_nome')}")
```

## 3. Modalidade + UF filter

```python
page = search(q="toner hp", modalidades=6, ufs="SP", tam_pagina=5)

print(f"Pregão Eletrônico / SP: {page['total_registros']}")
for r in page['items']:
    print(f"  {r.get('title','')[:80]} | {r.get('municipio_nome')}")
```

## 4. Raw response — inspect actual field names

```python
import httpx

resp = httpx.get(
    "https://pncp.gov.br/api/search/",
    params={"q": "toner", "tipos_documento": "edital", "pagina": 1, "tam_pagina": 1},
    headers={"User-Agent": "Mozilla/5.0"},
)
data = resp.json()
print("Top-level keys:", list(data.keys()))
if data.get("items"):
    print("\nFirst item keys:", list(data["items"][0].keys()))
    print("\nFirst item:")
    print(json.dumps(data["items"][0], indent=2, ensure_ascii=False))
```
